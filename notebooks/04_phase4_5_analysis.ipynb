# 04 — Phase 4.5 validation analysis
CPU/statistical aggregation only. This notebook rejects any input labelled TEST.

```python
from google.colab import drive
drive.mount('/content/drive')
import json, pathlib, numpy as np, pandas as pd
DRIVE_ROOT=pathlib.Path('/content/drive/MyDrive/evo2-distillation')
RUNS_ROOT=DRIVE_ROOT/'results/phase4_5'
run_dirs=[p for p in RUNS_ROOT.glob('*') if p.is_dir()]
for run in run_dirs:
    manifest=json.loads((run/'run_manifest.json').read_text())
    assert manifest.get('test_accessed') is False
    assert 'test' not in run.name.lower()
```

```python
summaries=[]
for run in run_dirs:
    metric=run/'metrics.json'
    if metric.exists(): summaries.append({'run_id':run.name,**json.loads(metric.read_text())})
summary=pd.DataFrame(summaries).sort_values('mean_within_genome_spearman',ascending=False) if summaries else pd.DataFrame()
display(summary)
```
